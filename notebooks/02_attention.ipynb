# Lab 02 · One-head attention

Queries, keys, and values on a 4-token sequence. Print the weights. They must sum to 1.

```python
import numpy as np

def softmax(x, axis=-1):
    x = x - x.max(axis=axis, keepdims=True)
    e = np.exp(x)
    return e / e.sum(axis=axis, keepdims=True)

rng = np.random.default_rng(0)
T, D = 4, 8
x = rng.normal(size=(T, D))
Wq = rng.normal(size=(D, D))
Wk = rng.normal(size=(D, D))
Wv = rng.normal(size=(D, D))
q, k, v = x @ Wq, x @ Wk, x @ Wv
scale = np.sqrt(D)
weights = softmax((q @ k.T) / scale)
out = weights @ v
assert weights.shape == (T, T)
assert np.allclose(weights.sum(axis=1), 1.0)
print("weights\n", np.round(weights, 3))
print("ok")
```
